```python
#poliformismo es la capacidad de un objeto de tomar muchas formas diferentes. 
#Por ejemplo podemos tener el método precio_iva, de la clase madre Producto, que calcule el precio aplicando el IVA:
class Producto:
        #... 
        def precio_iva(self):
                return self.precio * (1 + self.IVA)
```

```python
#Y queremos que aquellos productos que sean electrodomésticos se calcule ese precio con IVA pero restandole 100.
#Lo ´único que tendríamos que hacer es crear un método precio_iva en la clase Electrodomesticos que contemple esta resta:
class Electrodomestico(Producto):
        #... 
        def precio_iva(self):
                return self.precio * (1 + self.IVA) - 100
```
