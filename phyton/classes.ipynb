```python
#Lo primero que hacemos es definir las propiedades de nuestra clase, es decir los atributos. Todos los atributos deben estar definidos en un método llamado método constructor
#El método constructor se encuentra dentro de la clase y tiene por parámetros las características que tendrá nuestra clase siguiendo esta sintaxis:
class NombreClase :

	def __init__ (self, atributo1, atributo2, atributo3)
	
# Donde de __init__ es el método constructor
# self es el primer parámetro
# los atributos son las propiedades que queremos que tenga nuestra clase
```

```python
class Empleado:
    """
    Clase que representa a un empleado en una empresa.
    """

    def __init__(self, nombre, edad, posicion, salario):
        """
        Constructor de la clase Empleado.

        Args:
            nombre (str): El nombre del empleado.
            edad (int): La edad del empleado.
            posicion (str): La posición del empleado.
            salario (float): El salario del empleado.
        """
        # Creamos atributos y asignamos el valor de los parámetros del método constructor
        self.nombre = nombre
        self.edad = edad
        self.posicion = posicion
        self.salario = salario
```

```python
#Para instanciar un objeto de una clase que hayamos creado tenemos que seguir la siguiente sintaxis
#objeto = NombreClase(atributo1, atributo2, atributo3)
# Donde NombreClase es el nombre de nuestra clase
# Y los atributos son las propiedades recogidas en el método constructor
#un ejemplo con un objeto llamado juan de la clase Empleado, donde:"Juan" hace referencia al atributo nombre,30 al atributo edad,
# "Gerente" al atributo posicion.,50000 al atributo salario.
# Instanciamos un nuevo objeto(creamos un nuevo empleado)
juan = Empleado("Juan", 30, "Gerente", 50000)

# Mostramos por terminal los atributos de nuestro objeto juan
print(f"La posición de Juan es: {juan.posicion}")
print(f"Todos los atributos de Juan son: {juan.__dict__}")
```

```python
#Una vez tengamos el método constructor definimos las acciones que queremos que tenga nuestra clase, es decir, sus métodos propios. 
# Los métodos aparecen a continuación del método constructor.
# Por ejemplo un método que sirva para subir el salario según un porcentaje:
def subir_salario(self, porcentaje):
        """
        Método para aumentar el salario del empleado en un cierto porcentaje.

        Args:
            porcentaje (float): El porcentaje de aumento del salario.
        """
        # Calcula el nuevo salario aumentado por el porcentaje dado
        self.salario *= (1 + porcentaje / 100)
        # Redondea el salario a dos decimales
        self.salario = round(self.salario, 2)
        # Muestra un mensaje indicando que el salario ha subido y cuál es el nuevo salario
        return(f"¡El salario de {self.nombre} ha subido! Nuevo salario: {self.salario:.2f}")
        # Subimos el salario del empleado en un 10%
        
```
