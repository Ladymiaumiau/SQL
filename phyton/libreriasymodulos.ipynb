```python
#LIBRERIAS Y MODULOS
# Importamos la libreía para operaciones matemáticas
import math

# Definimos una variable y aplicamos el método sqrt() que calcula la raíz cuadrada
numero1 = 36
raiz_numero1 = math.sqrt(numero1)
print(f'La raiz cuadrada de nuestro número es ', raiz_numero1)
```

```python
#DATETIME
# Importamos la librería datetime para trabajar con fechas y horas  
 #el primer datetime hace referencia al módulo, el segundo al método llamado igual y los números dentro del paréntesis,
 # los argumentos del método, son el año, el mes, el día y la hora exacta.
import datetime

# Creamos un objeto datetime con una fecha y hora específicas. Año, mes, día, hora, min, segundo)
fecha_hora_especifica = datetime.datetime(1495, 2, 14, 12, 30, 0)
print("Fecha y hora específica:", fecha_hora_especifica)
# Output esperado: Fecha y hora específica: 1495-02-14 12:30:00
```

```python
#El método now() nos permite acceder a la fecha y hora exactas de nuestro ordenador:

# Creamos un objeto datetime con la fecha y hora actual
fecha_hora_actual = datetime.datetime.now()   
# Primero la librería.luego el método usado. now() otro método que hace referencia a ahora mismo

print("Fecha y hora actual:", fecha_hora_actual)
```

```python
#METODO DATE NOS PERMITE TRABAJAR SOLO CON FECHAS, SIN HORAS
# Creamos un objeto date con la fecha actual
fecha_hoy = datetime.date.today()
print("Fecha hoy: ", fecha_hoy)

# Creamos un objeto date con una fecha concreta
fecha_especifica = datetime.date(1495, 12, 31)
print("Fecha específica: ", fecha_especifica)
```

```python
#METODO TIME NOS PERMITE TRABAJAR SOLO CON HORAS, MINUTOS Y SEGUNDOS
# Creamos un objeto time con la hora

hora = datetime.datetime.now().time()
print("La hora es: ", hora)

hora_elegida = datetime.time(23, 14, 56)
print("La hora concreta es: ", hora_elegida)
```

```python
#METODO STRFTIME NOS PERMITE FORMATEAR FECHAS Y HORAS A NUESTRO GUSTO
# Obtenemos la fecha y hora actual
fecha_hora_actual = datetime.datetime.now()
print("La fecha sin formatear es :", fecha_hora_actual)

# Formateamos la fecha y hora en un formato específico
fecha_hora_formateada = fecha_hora_actual.strftime("%Y-%m-%d %H:%M:%S")
print("Fecha y hora formateada:", fecha_hora_formateada)

# Formateamos solo la fecha en otro formato
fecha_formateada = fecha_hora_actual.date().strftime("%d/%m/%Y")
print("Fecha formateada:", fecha_formateada)

# Formateamos solo la hora en otro formato
hora_formateada = fecha_hora_actual.time().strftime("%H:%M:%S")
print("Hora formateada:", hora_formateada)
```

```python
#METODO TIMEDELTA NOS PERMITE HACER OPERACIONES CON FECHAS Y HORAS
# Obtenemos la fecha actual
fecha_actual = datetime.date.today()
print("Fecha y hora actual:", fecha_hora_actual)

# Calculamos la fecha de hace una semana
fecha_hace_una_semana = fecha_actual - datetime.timedelta(days=7)
print("Fecha de hace una semana:", fecha_hace_una_semana)

# Calculamos la fecha de dentro de tres meses
fecha_dentro_de_tres_meses = fecha_actual + datetime.timedelta(days=90)
print("Fecha de dentro de tres meses:", fecha_dentro_de_tres_meses)
```

```python
#METODO READ() PARA LEER EL CONTENIDO DE UN FICHERO Y READLINE() PARA LEER UNA LINEA
#Usamos el método read() para leer todo el contenido de un fichero. Si queremos leer solo una línea usamos el método readline().
# Recuerda cambiar los argumentos por el nombre del archivo con el que vayas a trabajar

# Abrimos el fichero 'mi_fichero.txt' en modo de lectura
fichero_abrir = open('mi_fichero.txt', 'r')
# Leemos todo el contenido del fichero
contenido = fichero_abrir.read()
# Lee una línea del fichero
linea = fichero_abrir.readline()
```

```python
#METODO WRITE()
#Usamos el método write() para escribir en un fichero. Recuerda abrir el fichero en modo de escritura (w) o en modo de añadir (a) 
# en vez de en modo lectura(r) como hasta ahora.

# Abrimos el fichero 'mi_fichero.txt' en modo de escritura
fichero_escritura = open('mi_fichero.txt', 'w')

# Escribimos en el fichero
fichero_escritura.write("Hola, mundo!")
```

```python
#METODO CLOSE() PARA CERRAR FICHEROS ABIERTOS 
#Usamos el método close() para cerrar un fichero. Es importante cerrar los ficheros después de usarlos para liberar recursos del sistema.
# Abrimos un archivo con la función open
file = open('mi_fichero.txt', 'w)
# Cerramos el fichero
file.close()
```
